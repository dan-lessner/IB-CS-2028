```python
stack = []
stack.append(5)
stack.append(3)
print(stack) 
stack = []
expression = "5 3 + 2 *"

for x in expression.split():

    if x.isdigit():
        stack.append(int(x))

    elif x == "+":
        b = stack.pop()
        a = stack.pop()
        stack.append(a + b)

    elif x == "*":
        b = stack.pop()
        a = stack.pop()
        stack.append(a * b)

print(stack)
```
